## Comparación de métodos

Después de implementar y comparar ambos métodos, se concluye que el método de Runge-Kutta de cuarto orden (RK4) es más adecuado para simular el sistema de Lorenz. Esto se debe a que ofrece mayor precisión y estabilidad que el método de Euler, especialmente en un sistema tan sensible a las condiciones iniciales. Aunque RK4 requiere más cálculos por paso, permite trabajar con pasos más grandes sin perder fidelidad en los resultados.

En cambio, el método de Euler, al ser más simple y de menor orden, acumula errores rápidamente y puede generar trayectorias que no reflejan bien el comportamiento real del sistema, a menos que se utilice un paso de tiempo muy pequeño, lo que lo vuelve ineficiente.

Este comportamiento concuerda con lo discutido en la literatura, donde se explica que métodos de orden superior como RK4 logran mejores aproximaciones con menor error global y mayor estabilidad en problemas donde no se conoce una solución exacta [1].

[1] R. L. Burden y J. D. Faires, Numerical Analysis, 9th ed., Brooks/Cole, Cengage Learning, 2011, cap. 5.
